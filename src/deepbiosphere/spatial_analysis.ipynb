```python
## Change map tif and Variance decomposition tif
import os
import numpy as np
import rasterio
from rasterio.warp import reproject, Resampling
from deepbiosphere.Utils import paths


MERGED_DIR = f"{paths.RASTERS}invasive_future_merged_all/"
OUT_DIR = os.path.join(MERGED_DIR, "delta_anova_outputs")
os.makedirs(OUT_DIR, exist_ok=True)

SPECIES = ['Rhamnus_cathartica', 'Ailanthus_altissima']
GCMS = ['BCC-CSM2-MR', 'CMCC-ESM2', 'IPSL-CM6A-LR', 'MIROC6', 'MRI-ESM2-0']
SSPS = ['ssp126', 'ssp245', 'ssp585']

INC_THRESHOLD = 0.1
DEC_THRESHOLD = -0.1

CURRENT_TEMPLATE = (
    r"D:\deepbiosphere_mi\PATHS\RASTERS\invasive_current"
    r"\256m_2012_-1_mi_2012_-1_10\fully_merged\{spec}_current_masked.tif"
)

FUTURE_TEMPLATE = os.path.join(MERGED_DIR, "{spec}_{gcm}_{ssp}_merged_masked.tif")


def read_raster_as_float(path):
    with rasterio.open(path) as src:
        arr = src.read(1).astype(np.float32)
        profile = src.profile.copy()
        transform = src.transform
        crs = src.crs
        height = src.height
        width = src.width
        nodata = src.nodata

    if nodata is not None:
        arr[arr == nodata] = np.nan

    return arr, profile, transform, crs, height, width, nodata


def reproject_to_reference(src_path, ref_transform, ref_crs, ref_h, ref_w):
    with rasterio.open(src_path) as src:
        dst = np.full((ref_h, ref_w), np.nan, dtype=np.float32)

        reproject(
            source=rasterio.band(src, 1),
            destination=dst,
            src_transform=src.transform,
            src_crs=src.crs,
            dst_transform=ref_transform,
            dst_crs=ref_crs,
            src_nodata=src.nodata,
            dst_nodata=np.nan,
            resampling=Resampling.bilinear
        )

    return dst


def load_future_aligned(future_path, ref_transform, ref_crs, ref_h, ref_w):
    with rasterio.open(future_path) as src:
        needs_reproject = (
            src.crs != ref_crs or
            src.height != ref_h or
            src.width != ref_w or
            src.transform != ref_transform
        )

    if needs_reproject:
        fut = reproject_to_reference(future_path, ref_transform, ref_crs, ref_h, ref_w)
    else:
        fut, _, _, _, _, _, nodata = read_raster_as_float(future_path)
        if nodata is not None:
            fut[fut == nodata] = np.nan

    return fut.astype(np.float32)


def classify_change(mean_delta, inc_thres=0.1, dec_thres=-0.1):
    out = np.full(mean_delta.shape, 0, dtype=np.uint8)  # 0 = nodata
    valid = ~np.isnan(mean_delta)

    out[valid & (mean_delta < dec_thres)] = 1
    out[valid & (mean_delta >= dec_thres) & (mean_delta <= inc_thres)] = 2
    out[valid & (mean_delta > inc_thres)] = 3

    return out


def additive_ss_partition(y, gcm_ids, ssp_ids, n_gcm, n_ssp):
    if np.any(np.isnan(y)):
        return np.nan, np.nan, np.nan, np.nan

    grand_mean = np.mean(y)

    gcm_means = np.full(n_gcm, np.nan, dtype=np.float32)
    for g in range(n_gcm):
        gcm_means[g] = np.mean(y[gcm_ids == g])

    ssp_means = np.full(n_ssp, np.nan, dtype=np.float32)
    for s in range(n_ssp):
        ssp_means[s] = np.mean(y[ssp_ids == s])

    ss_total = np.sum((y - grand_mean) ** 2)

    ss_gcm = n_ssp * np.sum((gcm_means - grand_mean) ** 2)
    ss_ssp = n_gcm * np.sum((ssp_means - grand_mean) ** 2)

    fitted = np.empty_like(y, dtype=np.float32)
    for i in range(len(y)):
        fitted[i] = gcm_means[gcm_ids[i]] + ssp_means[ssp_ids[i]] - grand_mean

    ss_resid = np.sum((y - fitted) ** 2)

    return ss_gcm, ss_ssp, ss_resid, ss_total


def write_raster(out_path, arr, ref_profile, dtype, nodata):
    profile = ref_profile.copy()
    profile.update(
        dtype=dtype,
        count=1,
        compress='lzw',
        nodata=nodata
    )
    with rasterio.open(out_path, 'w', **profile) as dst:
        dst.write(arr, 1)


for spec in SPECIES:
    print(f"\n{'='*70}")
    print(f"Processing species: {spec}")
    print(f"{'='*70}")

    current_path = CURRENT_TEMPLATE.format(spec=spec)
    current_data, ref_profile, ref_transform, ref_crs, ref_h, ref_w, _ = read_raster_as_float(current_path)

    n_gcm = len(GCMS)
    n_ssp = len(SSPS)
    n_scen = n_gcm * n_ssp

    delta_stack = np.full((n_scen, ref_h, ref_w), np.nan, dtype=np.float32)

    gcm_ids = []
    ssp_ids = []

    idx = 0
    for g_i, gcm in enumerate(GCMS):
        for s_i, ssp in enumerate(SSPS):
            future_path = FUTURE_TEMPLATE.format(spec=spec, gcm=gcm, ssp=ssp)
            print(f"  Loading {gcm}_{ssp}")

            future_data = load_future_aligned(
                future_path, ref_transform, ref_crs, ref_h, ref_w
            )

            valid = (~np.isnan(current_data)) & (~np.isnan(future_data))
            delta = np.full((ref_h, ref_w), np.nan, dtype=np.float32)
            delta[valid] = future_data[valid] - current_data[valid]

            delta_stack[idx, :, :] = delta
            gcm_ids.append(g_i)
            ssp_ids.append(s_i)
            idx += 1

    gcm_ids = np.array(gcm_ids, dtype=int)
    ssp_ids = np.array(ssp_ids, dtype=int)

    mean_delta = np.nanmean(delta_stack, axis=0)
    class_map = classify_change(mean_delta, INC_THRESHOLD, DEC_THRESHOLD)

    class_out = os.path.join(OUT_DIR, f"{spec}_delta_class_map.tif")
    write_raster(class_out, class_map, ref_profile, dtype='uint8', nodata=0)
    print(f" Saved class map: {class_out}")

    gcm_prop_map = np.full((ref_h, ref_w), np.nan, dtype=np.float32)
    ssp_prop_map = np.full((ref_h, ref_w), np.nan, dtype=np.float32)
    resid_prop_map = np.full((ref_h, ref_w), np.nan, dtype=np.float32)
    dom_source_map = np.zeros((ref_h, ref_w), dtype=np.uint8)


    all_valid = np.all(~np.isnan(delta_stack), axis=0)
    valid_positions = np.argwhere(all_valid)

    print(f"  Valid pixels for ANOVA decomposition: {len(valid_positions)}")

    for count, (r, c) in enumerate(valid_positions, start=1):
        y = delta_stack[:, r, c]

        ss_gcm, ss_ssp, ss_resid, ss_total = additive_ss_partition(
            y, gcm_ids, ssp_ids, n_gcm, n_ssp
        )

        if np.isnan(ss_total) or ss_total <= 0:
            continue

        g_prop = ss_gcm / ss_total
        s_prop = ss_ssp / ss_total
        r_prop = ss_resid / ss_total

        gcm_prop_map[r, c] = g_prop
        ssp_prop_map[r, c] = s_prop
        resid_prop_map[r, c] = r_prop

        dominant = np.argmax([g_prop, s_prop, r_prop]) + 1
        dom_source_map[r, c] = dominant

        if count % 50000 == 0:
            print(f" processed {count}/{len(valid_positions)} pixels")

    gcm_out = os.path.join(OUT_DIR, f"{spec}_gcm_ss_prop.tif")
    ssp_out = os.path.join(OUT_DIR, f"{spec}_ssp_ss_prop.tif")
    resid_out = os.path.join(OUT_DIR, f"{spec}_resid_ss_prop.tif")
    dom_out = os.path.join(OUT_DIR, f"{spec}_dominant_uncertainty_source.tif")

    write_raster(gcm_out, gcm_prop_map, ref_profile, dtype='float32', nodata=np.nan)
    write_raster(ssp_out, ssp_prop_map, ref_profile, dtype='float32', nodata=np.nan)
    write_raster(resid_out, resid_prop_map, ref_profile, dtype='float32', nodata=np.nan)
    write_raster(dom_out, dom_source_map, ref_profile, dtype='uint8', nodata=0)

    print(f" Saved GCM proportion map: {gcm_out}")
    print(f" Saved SSP proportion map: {ssp_out}")
    print(f" Saved residual proportion map: {resid_out}")
    print(f" Saved dominant source map: {dom_out}")

print("\nAll done.")
```

```python
## single change map and variance source maps
import os
import numpy as np
import rasterio
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap, BoundaryNorm
from matplotlib.patches import Patch
from deepbiosphere.Utils import paths

MERGED_DIR = f"{paths.RASTERS}invasive_future_merged_all/"
IN_DIR = os.path.join(MERGED_DIR, "delta_anova_outputs")
FIG_DIR = os.path.join(IN_DIR, "figures")
os.makedirs(FIG_DIR, exist_ok=True)

SPECIES = ['Rhamnus_cathartica', 'Ailanthus_altissima']

def read_raster(path):
    with rasterio.open(path) as src:
        arr = src.read(1)
        nodata = src.nodata
    arr = arr.astype(np.float32)
    if nodata is not None:
        arr[arr == nodata] = np.nan
    return arr

def save_map(fig, outpath):
    fig.savefig(outpath, dpi=300, bbox_inches='tight')
    plt.close(fig)
    print(f"Saved: {outpath}")

def plot_change_class_map(spec):
    path = os.path.join(IN_DIR, f"{spec}_delta_class_map.tif")
    arr = read_raster(path)

    cmap = ListedColormap([
        "#4daf4a",  # 1 decrease = green
        "#377eb8",  # 2 stable   = blue
        "#e41a1c"   # 3 increase = red
    ])
    norm = BoundaryNorm([0.5, 1.5, 2.5, 3.5], cmap.N)

    fig, ax = plt.subplots(figsize=(8, 6))
    ax.imshow(arr, cmap=cmap, norm=norm)
    ax.axis("off")

    legend_elements = [
        Patch(facecolor="#4daf4a", label='Decrease'),
        Patch(facecolor="#377eb8", label='Stable'),
        Patch(facecolor="#e41a1c", label='Increase'),
    ]
    ax.legend(
        handles=legend_elements,
        loc='lower left',
        frameon=True,
        fontsize=8
    )

    out = os.path.join(FIG_DIR, f"{spec}_A_change_class_map.png")
    save_map(fig, out)

def plot_dominant_source_map(spec):
    path = os.path.join(IN_DIR, f"{spec}_dominant_uncertainty_source.tif")
    arr = read_raster(path)

    cmap = ListedColormap([
        "#4daf4a",  # 1 GCM
        "#984ea3",  # 2 SSP
        "#ff7f00"   # 3 Residual
    ])
    norm = BoundaryNorm([0.5, 1.5, 2.5, 3.5], cmap.N)

    fig, ax = plt.subplots(figsize=(8, 6))
    ax.imshow(arr, cmap=cmap, norm=norm)
    ax.axis("off")

    legend_elements = [
        Patch(facecolor="#4daf4a", label='GCM dominated'),
        Patch(facecolor="#984ea3", label='SSP dominated'),
        Patch(facecolor="#ff7f00", label='Residual dominated'),
    ]
    ax.legend(
        handles=legend_elements,
        loc='lower left',
        frameon=True,
        fontsize=6
    )

    out = os.path.join(FIG_DIR, f"{spec}_B_dominant_source_map.png")
    save_map(fig, out)

def plot_contribution_maps(spec):
    gcm_path = os.path.join(IN_DIR, f"{spec}_gcm_ss_prop.tif")
    ssp_path = os.path.join(IN_DIR, f"{spec}_ssp_ss_prop.tif")
    resid_path = os.path.join(IN_DIR, f"{spec}_resid_ss_prop.tif")
    gcm_arr = read_raster(gcm_path)
    ssp_arr = read_raster(ssp_path)
    resid_arr = read_raster(resid_path)

    vmin = 0.0
    vmax = 1.0

    fig, axes = plt.subplots(1, 3, figsize=(22, 6))

    im1 = axes[0].imshow(gcm_arr, cmap='viridis', vmin=vmin, vmax=vmax)
    axes[0].set_title("GCM Contribution")
    axes[0].axis("off")

    im2 = axes[1].imshow(ssp_arr, cmap='viridis', vmin=vmin, vmax=vmax)
    axes[1].set_title("SSP Contribution")
    axes[1].axis("off")

    im3 = axes[2].imshow(resid_arr, cmap='viridis', vmin=vmin, vmax=vmax)
    axes[2].set_title("Residual Contribution")
    axes[2].axis("off")

    cbar = fig.colorbar(im2, ax=axes.ravel().tolist(), shrink=0.82, pad=0.02)
    cbar.set_label("Proportion of Sum of Squares")

    out = os.path.join(FIG_DIR, f"{spec}_C_contribution_maps.png")
    save_map(fig, out)


for spec in SPECIES:
    print(f"\n--- Plotting maps for {spec} ---")
    plot_change_class_map(spec)       # A
    plot_dominant_source_map(spec)    # B
    plot_contribution_maps(spec)      # C

print("\n maps done.")
```

```python
## change and dominant source csv summary
import os
import numpy as np
import pandas as pd
import rasterio
import matplotlib.pyplot as plt
from deepbiosphere.Utils import paths

MERGED_DIR = f"{paths.RASTERS}invasive_future_merged_all/"
IN_DIR = os.path.join(MERGED_DIR, "delta_anova_outputs")
FIG_DIR = os.path.join(IN_DIR, "figures_stats")
os.makedirs(FIG_DIR, exist_ok=True)

SPECIES = ['Rhamnus_cathartica', 'Ailanthus_altissima']
SPECIES_LABELS = {
    'Rhamnus_cathartica': 'Rhamnus cathartica',
    'Ailanthus_altissima': 'Ailanthus altissima'
}

def read_raster(path):
    with rasterio.open(path) as src:
        arr = src.read(1).astype(np.float32)
        nodata = src.nodata
    if nodata is not None:
        arr[arr == nodata] = np.nan
    return arr

change_rows = []
source_rows = []
violin_rows = []

for spec in SPECIES:
    # ---------- 1. Change class map ----------
    class_path = os.path.join(IN_DIR, f"{spec}_delta_class_map.tif")
    class_arr = read_raster(class_path)

    valid_class = ~np.isnan(class_arr)
    total_valid_class = np.sum(valid_class)

    # 1=decrease, 2=stable, 3=increase
    for code, label in zip([1, 2, 3], ['Decrease', 'Stable', 'Increase']):
        count = np.sum(class_arr[valid_class] == code)
        prop = count / total_valid_class if total_valid_class > 0 else np.nan
        change_rows.append({
            'species': SPECIES_LABELS[spec],
            'category': label,
            'count': int(count),
            'proportion': prop
        })

    dom_path = os.path.join(IN_DIR, f"{spec}_dominant_uncertainty_source.tif")
    dom_arr = read_raster(dom_path)

    valid_dom = ~np.isnan(dom_arr)
    total_valid_dom = np.sum(valid_dom)

    for code, label in zip([1, 2, 3], ['GCM', 'SSP', 'Residual']):
        count = np.sum(dom_arr[valid_dom] == code)
        prop = count / total_valid_dom if total_valid_dom > 0 else np.nan
        source_rows.append({
            'species': SPECIES_LABELS[spec],
            'source': label,
            'count': int(count),
            'proportion': prop
        })

    gcm_path = os.path.join(IN_DIR, f"{spec}_gcm_ss_prop.tif")
    ssp_path = os.path.join(IN_DIR, f"{spec}_ssp_ss_prop.tif")
    resid_path = os.path.join(IN_DIR, f"{spec}_resid_ss_prop.tif")

    gcm_arr = read_raster(gcm_path)
    ssp_arr = read_raster(ssp_path)
    resid_arr = read_raster(resid_path)

    for arr, label in zip(
        [gcm_arr, ssp_arr, resid_arr],
        ['GCM', 'SSP', 'Residual']
    ):
        vals = arr[~np.isnan(arr)].ravel()

        max_points = 100000
        if len(vals) > max_points:
            rng = np.random.default_rng(42)
            vals = rng.choice(vals, size=max_points, replace=False)

        temp_df = pd.DataFrame({
            'species': SPECIES_LABELS[spec],
            'source': label,
            'value': vals
        })
        violin_rows.append(temp_df)

df_change = pd.DataFrame(change_rows)
df_source = pd.DataFrame(source_rows)
df_violin = pd.concat(violin_rows, ignore_index=True)

df_change.to_csv(os.path.join(FIG_DIR, "change_class_summary.csv"), index=False)
df_source.to_csv(os.path.join(FIG_DIR, "dominant_source_summary.csv"), index=False)
df_violin.to_csv(os.path.join(FIG_DIR, "violin_contribution_values_sampled.csv"), index=False)

print("Summary tables saved.")
```
