```python
import glob
import time
import numpy as np
import deepbiosphere.NAIP_Utils as naip
from deepbiosphere.Utils import paths

SPECIES = ['Rhamnus_cathartica', 'Ailanthus_altissima']

SCENARIOS = [
    'invasive_BCC-CSM2-MR_ssp126',
    'invasive_BCC-CSM2-MR_ssp245',
    'invasive_BCC-CSM2-MR_ssp585',
    'invasive_CMCC-ESM2_ssp126',
    'invasive_CMCC-ESM2_ssp245',
    'invasive_CMCC-ESM2_ssp585',
    'invasive_IPSL-CM6A-LR_ssp126',
    'invasive_IPSL-CM6A-LR_ssp245',
    'invasive_IPSL-CM6A-LR_ssp585',
    'invasive_MIROC6_ssp126',
    'invasive_MIROC6_ssp245',
    'invasive_MIROC6_ssp585',
    'invasive_MRI-ESM2-0_ssp126',
    'invasive_MRI-ESM2-0_ssp245',
    'invasive_MRI-ESM2-0_ssp585',
]

SAVE_DIR = '256m_2012_-1_mi_2012_-1_10'

for scenario in SCENARIOS:
    print(f"\n{'='*50}")
    print(f"{scenario}")
    BASE = f"{paths.RASTERS}{scenario}/{SAVE_DIR}/"

    for spec in SPECIES:
        start = time.time()
        files = glob.glob(f"{BASE}**/probability_{spec}.tif", recursive=True)
        files = [f for f in files if 'fully_merged' not in f and 'merging_temp' not in f]
        print(f"  {spec}: {len(files)} tiles")

        if len(files) == 0:
            continue

        naip.merge_lots_of_tiffs(
            parent_dir=scenario,
            save_dir=SAVE_DIR,
            pred_type='PER_SPEC',
            files=files,
            spec=spec,
            band_names=[spec]
        )

print("\nFinished")
```
